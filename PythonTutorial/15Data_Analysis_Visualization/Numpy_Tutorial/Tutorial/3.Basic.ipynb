### Broadcasting = It is a mechanism in which numpy perform operations on the array of different shape without using loop.
Rules = 
1. Either elements of the rows need to be same or any of them is 1.
2. You can easily do broadcasting of the vector and scaler
3. You can easily do broadcasting of the vector and vector
4. You can easily do broadcasting of the vector and matrix
5. You can easily do broadcasting of the matrix and matrix

### Stacking
It simply means joining the multiple array along a new  or existing axis.
It always return copy.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
x = np.array([[1, 2],
              [3, 4]])
y = np.array([[5, 6],
              [7, 8]])

print(np.hstack((a,b)))  # it join the arrays horizontally.
print(np.hstack((x,y)))  # it join the arrays horizontally.
print(np.vstack((a,b)))  # it join the arrays vertically.
print(np.vstack((x,y)))  # it join the arrays vertically.

print(np.column_stack((a,b)))
print(np.column_stack((x,y)))
print(np.row_stack((a,b)))
print(np.row_stack((x,y)))

# np.vsplit(arr,number of split) = Used to split the array into given numbers of equal parts in rows.If possible else not.
# np.hsplit(arr,number of split) = Used to split the array into given numbers of equal parts in columns.If possible else not.
```
