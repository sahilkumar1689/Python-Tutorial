### Basic Practive Question:

```python
import numpy as np
```

```python
# 1. Create a NumPy array from [1, 2, 3, 4, 5].
# Print its shape, dtype, and ndim.

first_arr = np.array([1,2,3,4,5]);
print(first_arr.shape," ",first_arr.dtype," ",first_arr.ndim," ",first_arr.size);
```

```python
# 2. Create a 3×3 array of zeros and a 3×3 array of ones.
# Add them together.

zero_arr = np.zeros((3,3),dtype=np.int32);
ones_arr = np.ones((3,3),dtype=np.int32);

print(zero_arr+ones_arr);
# print(np.add(zero_arr,ones_arr))
```

```python
# 3. From array [10, 20, 30, 40, 50]:
# extract the first, last, and middle element.

thirs_arr = np.array([10, 20, 30, 40, 50])
print("First= ",thirs_arr[0])
print("middle= ",thirs_arr[int(thirs_arr.size/2)])
print("Last= ",thirs_arr[-1])
```

```python
# 4. From array [1, 2, 3, 4, 5, 6]:
# get elements at even indices.

four_arr = np.array([1, 2, 3, 4, 5, 6]);

print(four_arr[four_arr %2 ==0])

```

```python
# 5. From [5, 12, 8, 20, 3]:
# extract values greater than 10.

five_arr = np.array([5, 12, 8, 20, 3])
print(five_arr[five_arr > 10])

# 6. Reshape array [1, 2, 3, 4, 5, 6] into a 2×3 matrix.
six_arr = np.array([1, 2, 3, 4, 5, 6]);
print(six_arr.reshape(2,3))

# 7.Compute square root of [1, 4, 9, 16].

print(np.sqrt([1,4,9,16]))

# 8. Find unique elements from [1, 2, 2, 3, 3, 3].

print(np.unique_values([1, 2, 2, 3, 3, 3]))

# 9. Sort array [3, 1, 4, 2] in ascending order.
print(np.sort([3,1,4,2]))

# 10. Add 5 to every element of array [1, 2, 3, 4].

print(np.array([1,2,3,4])+5)
```

```python
# 1. extract the second row
# extract the third column

md_1 = np.array([[10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]])

print(md_1[1])
print(md_1[:,1])

# 2. Extract the top-left 2×2 sub-matrix.

print(md_1[:2,:2])

# 3. Replace all values less than 50 with 0.

print(np.where(md_1 <50,0,md_1))

# 4. replace values > 50 with 1, else 0.

print(np.where(md_1 >50,1,0))

```

### Hard

```python
# 1. Normalize array [10, 20, 30, 40, 50] to range 0–1.
# Formula = (array-min)/(max-min)
 

h1 = np.array([10, 20, 30, 40, 50]);
min_val = np.min(h1)
max_val = np.max(h1)
print((h1-min_val)/(max_val-min_val))

# 2. Columns wise std:
h2 = np.array([[1, 2],
 [3, 4],
 [5, 6]]);

print(np.std(h2,axis=0))
print(np.mean(h2,axis=1))

# 3. Replace all odd numbers wit -1:

print(np.where(h2 %2 != 0, -1,h2))

# 4. matrix multiplication

mat_1 = np.array([[1, 2],
 [3, 4]])

mat_2 = np.array([[5, 6],
 [7, 8]])

print(np.matmul(mat_1,mat_2))

# 5. Solve this equation :
# 2x + y = 5
# x + 3y = 6

eq_1 = np.array([[2,1],[1,3]]);
eq_2 = np.array([5,6]);

print(np.linalg.solve(eq_1,eq_2));

# 6. Sort by column:
sort_arr = np.array([[1, 3],
 [2, 1],
 [3, 2]]);

print(np.sort(sort_arr,axis=0))

# 7. From [1, 2, 2, 3, 4, 4, 5]:

# find elements that appear more than once.

dup_arr = np.array([1, 2, 2, 3, 4, 4, 5]);
first,second= np.unique(dup_arr,return_counts=True)
print(first," ",second)
print(first[second>1])

# 8. find average per student

# find topper (highest total score)

std_marks = np.array([[85, 90, 88],
 [78, 75, 80],
 [92, 95, 94]])

print(np.mean(std_marks,axis=1))
print(np.max(np.sum(std_marks,axis=1)))



```

```python
# 1. Average marks per student

# Average marks per subject

# Identify the top performer

# Find students scoring below subject average

ad_1 = np.array([[78, 85, 90],
 [88, 76, 92],
 [91, 89, 84],
 [65, 70, 72]])

std_avg = np.mean(ad_1,axis=1)
print(std_avg)
sub_avg = np.mean(ad_1,axis=0)
print(sub_avg)

print(np.max(std_avg))
```

```python
# 4. Apply 10% discount

# Replace prices > 2000 with 2000

# Round final prices


h4 = np.array([999, 1499, 1999, 2499, 2999]);
dis_arr = h4*(1/10)
after_dis = h4-dis_arr
print(after_dis)
print(np.round(np.where(after_dis>2000,2000,after_dis)))

# 5. Replace 0 with column average

# Compute average rating per user

h5 = np.array([[5, 4, 0],
 [3, 0, 2],
 [4, 5, 3]])

col_avg = np.mean(h5,axis=0)

print(col_avg)

print(np.where(h5 == 0,col_avg,h5))


```
